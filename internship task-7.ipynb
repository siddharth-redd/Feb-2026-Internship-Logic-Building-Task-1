#Problem 1: Smart Light Controller

Problem Statement
Create a class to control a smart light that can be turned ON or OFF and display its status.

Sample Input
Light Name: Bedroom Light

Expected Output
Bedroom Light is ON


```python
class SmartLight:
    def __init__(self):
        self.light_name = None
        self.status = None

    def capture_light(self):
        self.light_name = input("Light Name: ")
        self.status = input("ON/OFF: ")

    def display_status(self):
        print(self.light_name + " is " + self.status)

light_controller = SmartLight()
light_controller.capture_light()
light_controller.display_status()
```

#Problem 2: Employee ID Card System

Problem Statement
Create a class that stores employee name, ID, and department and displays ID card details.
    
Sample Input:

Employee Name: Rahul

Employee ID: EMP102

Department: IT

Expected Output:
    
Employee ID Card
    
Name: Rahul
    
ID: EMP102
    
Department: IT

```python
class Employee_detail:
    def __init__(self):
        self.name = None
        self.emp_id = None
        self.department = None

    def capture_employee(self):
        self.name = input("Employee Name: ")
        self.emp_id = input("Employee ID: ")
        self.department = input("Department: ")

    def print_id_card(self):
        print("Employee ID Card")
        print("Name: " + self.name)
        print("ID: " + self.emp_id)
        print("Department: " + self.department)

emp1 = Employee_detail()
emp1.capture_employee()
emp1.print_id_card()
```

#Problem 3: Mobile Contact Record

Problem Statement
Create a class to store contact name and phone number and display contact information.

Contact Name: Anita

```python
class ContactRecord:
    def __init__(self):
        self.contact_name = None
        self.phone_number = None

    def capture_contact(self):
        self.contact_name = input("Contact Name: ")
        self.phone_number = input("Phone Number: ")

    def display_contact(self):
        print("Contact Saved")
        print("Name: " + self.contact_name)
        print("Phone: " + self.phone_number)

c1 = ContactRecord()
c1.capture_contact()
c1.display_contact()
```
