#Problem 1: Unique Words in a Sentence

Given a sentence, identify all unique words and display:
1) The count of unique words
2) The unique words themselves

   
Input
Enter sentence: python is easy and python is powerful


```python
inp= "python is easy and python is powerful"
x=set(inp.split(" "))

count=0
for i in x:
    count=count+1

print("Unique words count: ",count)
print("Unique words: ",x)
```

Problem 2: Highest Salary from Employee Data

Store employee names and their salaries, and determine which employee has the highest salary.


```python
employees = {"Ravi": 75000,"Anita": 68000,"Kiran": 72000}
x=list(employees.items())
highest=0
highest_name=" "
for i in x:
    if i[1]>highest:
        highest=i[1]
        highest_name=i[0]

print("highest salary: ",highest_name," - ",highest)
```

Problem 3: Find Maximum and Minimum Values

Find the maximum and minimum values from a list of numbers.

```python
x=[45, 22, 89, 10, 66]
high=0
low=1000
for i in x:
    if i>high:
        high=i
    elif i<low:
        low=i
print("List: ",x)
print("Max: ",high)
print("Min: ",low)
```

Problem 4: Count Products Above a Price Threshold

Count how many products have a price greater than 1000.

Problem 5: Calculate Attendance Percentage

Calculate the attendance percentage from a list containing "P" (Present) and "A" (Absent).


```python
attendance = ["P", "P", "A", "P", "P"]
count=0
for i in attendance:
    if i=="P":
        count=count+1
att_percentage=(count/len(attendance))*100
print("attendance percentage: ",att_percentage)
```

Problem 7: Count Character Frequency

Count how many times each character appears in a given string.


```python
x="python"
y=[]
freq = {}

for i in x:
    if i in y:
        freq[i]=freq[i]+1
    else:
        y.append(i)
        freq[i] = 1

print(freq)
```

Problem 9: Check if a Key Exists in a Dictionary

Check whether a specific key exists in a dictionary.


```python
def check(x,key):
    for i in x.keys():
        if i==key:
            return i+" exists"
x={"employee":1,"student":2,"teacher":3}
check(x,"employee")
```
