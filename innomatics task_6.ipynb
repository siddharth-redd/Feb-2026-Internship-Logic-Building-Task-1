#Problem Statement 1:Smart Parking Lot Management System

Design a function to manage a smart parking lot.

The system should:

1) Accept vehicle entry and exit logs

2) Calculate total parked vehicles

3) Identify peak parking usage

4) Alert if parking exceeds capacity

```python
def smart_parking(capacity, vehicle_logs):
    current = 0
    peak = 0
    alert = False

    for i in vehicle_logs:
        if i == "IN":
            current += 1
        elif i == "OUT":
            if current > 0:
                current -= 1
        
        if current > peak:
            peak = current
        
        if current > capacity:
            alert = True

    print("Currently Parked Vehicles:", current)
    print("Peak Parking Usage:", peak)
    
    if alert:
        print("Alert: Parking capacity exceeded")
    else:
        print("Parking Status: Available")

capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, logs)
```

#Problem Statement 2:Online Food Delivery Time Estimator

Create a function that estimates delivery time based on:

1) Distance (km)

2) Weather condition

3) Traffic level


```python
def estimate_delivery_time(distance, traffic, weather):
    total_time = distance * 5 #5mins per km
    
    if traffic == "Low":
        total_time += 5
    elif traffic == "Medium":
        total_time += 10
    elif traffic == "High":
        total_time += 15
    
    if weather == "Rainy":
        total_time += 10
    elif weather == "Stormy":
        total_time += 20
    
    print("Estimated Delivery Time: ",total_time)

estimate_delivery_time(8,"high","rainy")
```

#Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

Build a function that analyzes seat booking data and:

1) Calculates occupancy percentage

2) Determines if show is Housefull

3) Suggests opening additional shows


```python
def seat_occupancy(total_seats, booked_seats):
    occupancy = (len(booked_seats) / total_seats) * 100
    print("Occupancy:",occupancy, "%")
    
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
        print("Suggestion: Consider opening additional shows.")
    else:
        print("Show Status: Seats Available")

total=200
booked_seats=[1]*150
seat_occupancy(total,booked_seats)
```

#Problem Statement 4:Cloud Server Load Classification System

Create a function to classify server load based on CPU usage readings.

Rules:

1) Average CPU < 50% → Normal

2) 50%–80% → Warning

3) 80% → Critical


```python
def server_load(x):
    avg=sum(x)/len(x)
    if avg <= 50:
        print("average CPU load: ",avg)
        print("server status: Normal")
    elif ((avg > 50) and (avg <= 80)):
        print("average CPU load: ",avg)
        print("server status: Warning")
    else:
        print("average CPU load: ",avg)
        print("server status: Critical")

x=[45, 60, 70, 85, 90]
server_load(x)
```

#Problem Statement 5:Smart Classroom Resource Usage Monitor

Design a function that tracks usage of classroom resources (projector, AC, lights) and identifies overuse patterns.


```python
resource_usage={"Projector": 6,"AC": 9,"Lights": 4}

def monitor_resources(resource_usage):
    overused = []
    threshold = 8  
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused.append(resource)
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

monitor_resources(resource_usage)
```

#Problem Statement 6: Online Event Registration Capacity Controller
 
Create a function that manages event registrations by:

Tracking registrations

Preventing overbooking

Triggering waitlist mode


```python
def manage_registrations(capacity, registrations):
    
    if registrations <= capacity:
        confirmed = registrations
        waitlist = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)

capacity = 100
registrations = 105
manage_registrations(capacity,registrations)
```
