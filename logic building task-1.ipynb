**1.User Login Check**

Problem Statement: Given a username and password, check whether login is successful.

username = "admin"
password = "1234"

Requirements: Print "Login Successful" if both username and password match Otherwise print "Invalid Credentials"

```python
username = "admin"
password = "1234"

user_name = "admin"
pass_word = "1234"

if(( username==user_name) and (password==pass_word)): #condition to check if username and password are matching
    print("Login Successful")
else:
    print("Invalid Credentials")
```

**2. Pass / Fail Analyzer**

Problem Statement:
Analyze student results.
marks = [45, 78, 90, 33, 60]

Requirements:
A student passes if marks ≥ 50
Count the total number of pass students
Count the total number of fail students
Print the final result clearly

Real-World Application: Academic evaluation systems


```python
marks = [45, 78, 90, 33, 60]
pass_count=0
fail_count=0
for i in marks:
    if i<50:
        fail_count=fail_count+1   #counting number of failed students if marks<50
    else:
        pass_count=pass_count+1   #counting number of passed students if marks>50
print("passed:",pass_count)
print("failed:",fail_count)
```

**3. Simple Data Cleaner**

Problem Statement:
Clean and standardize user names.
names = [" Alice ", "bob", " CHARLIE "]

Requirements:
Remove extra spaces
Convert all names to lowercase
Print the cleaned list

Real-World Application: Data preprocessing before analysis


```python
names = [" Alice ", "bob", " CHARLIE "]
cleaned_name_list=[]
for i in names:                                    #lower is used to convert the all strings to lowercase
    cleaned_name_list.append((i.strip()).lower())  #strip is used to remove all leading and trailling extra spaces
cleaned_name_list                                  #finally we append the cleaned strings into a new list   
```

**4. Message Length Analyzer**

Problem Statement:
Analyze message sizes.
messages = ["Hi", "Welcome to the platform", "OK"]

Requirements:
Print the length of each message
Flag messages longer than 10 characters

Real-World Application: Text filtering and validation systems


```python
messages = ["Hi", "Welcome to the platform", "OK"]
for i in messages:
    if len(i)<10:                              #checking if message length<10
        print(i,":",len(i))
    else:
        if(len(i)>10):
            print(i,":",len(i),"(flagged)")    #checking if message length>10
```

**5. Error Message Detector**
    
Problem Statement:
Detect error messages from system logs.
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

Requirements:
Count the number of "ERROR" entries


```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]
count=0
for i in logs:
    if i=="ERROR":                            #condition to check if the messages == "ERROR"
        count=count+1
print("number of ERROR entries: ",count)
```
